```python
x1 = float(input('첫 번째 지점의 x좌점:'))
y1 = float(input('첫 번째 지점의 y좌표:'))
x2 = float(input('두 번째 지점의 x좌표:'))
y2 = float(input('두 번째 지점의 y좌표:'))
distance = ((x2-x1)**2+(y2-y1)**2)**0.5
print(f'두 지점 사이의 거리는 {distance:2f}입니다.')
```
